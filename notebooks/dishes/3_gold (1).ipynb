# Генерація дієтичних тегів
створення словника з забороненими словами та проходимося по ньому 

```python
diet_rules = {
    'lactose_intolerant': ['milk', 'cheese', 'butter', 'cream', 'yogurt', 'whey', 'parmesan', 'mozzarella'],
    'vegetarian': ['beef', 'pork', 'chicken', 'fish', 'bacon', 'meat', 'ham', 'sausage', 'turkey', 'shrimp'],
    'vegan': ['beef', 'pork', 'chicken', 'fish', 'bacon', 'meat', 'ham', 'sausage', 'turkey', 'shrimp', 
              'milk', 'cheese', 'butter', 'cream', 'yogurt', 'eggs', 'egg', 'honey'],
    'nut_allergy': ['peanut', 'almond', 'walnut', 'pecan', 'macadamia', 'nuts', 'cashew', 'pine nut'],
    'gluten_free': ['flour', 'wheat', 'bread', 'pasta', 'macaroni', 'spaghetti', 'noodle'],
    'no_tomatoes': ['tomato', 'ketchup'], 
    'no_onions_garlic': ['onion', 'garlic'] # "onion" зловить "yellow onions", "green onions"
}

def get_suitable_diets(ing_str):
    try:
        ingredients_list = ast.literal_eval(ing_str)
    except:
        ingredients_list = []

    suitable = []
    
    for diet, forbidden_foods in diet_rules.items():
        is_suitable = True
        
        for ingredient in ingredients_list:
            ingredient = ingredient.lower()
            for forbidden in forbidden_foods:
                if forbidden in ingredient:
                    is_suitable = False
                    break 
            if not is_suitable:
                break 
                
        if is_suitable:
            suitable.append(diet)
            
    return suitable

dishes_gold['suitable_diets'] = dishes_gold['ingredients'].apply(get_suitable_diets)

final_dishes = dishes_gold.copy()

display(final_dishes[['name', 'ingredients', 'suitable_diets']].head(5))
```
